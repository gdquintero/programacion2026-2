```python
def suma_vectores(x,y,res):
    res[:] = x[:] + y[:]
    return res

def producto_escalar(x,alpha,res):
    res[:] = alpha * x[:]
    return res

def producto_interno(x,y):
    res = 0

    for i in range(len(x)):
        res = res + x[i] * y[i]

    return res

def producto_externo(x,y,res):
    for i in range(len(x)):
        res[i,:] = x[i] * y[:]

    return res

x = np.array([1,2,3,4])
y = np.array([2,1,3,2])

vec_aux = np.zeros(len(x))
mat_aux = np.zeros((len(x),len(y)))

# print(suma_vectores(x,y,vec_aux))

# print(producto_escalar(x,2,vec_aux))

# print(producto_interno(x,y))

print(producto_externo(x,y,mat_aux))
```
